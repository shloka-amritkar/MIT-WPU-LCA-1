```python
### LISTS 5 OPERATIONS ###
```

```python
students = ["Rutuj", "Rodge", "Yash", "Shloka"]
students.append("Aditya")
print(students)
```

```python
students = ["Rutuj", "Rodge", "Yash", "Shloka"]
students.insert(1, "Aditya")
print(students)
```

```python
students = ["Rutuj", "Rodge", "Yash", "Shloka"]
students.remove("Yash")
print(students)
```

```python
students = ["Rutuj", "Rodge", "Yash", "Shloka"]
students.pop(2)
print(students)
```

```python
students = ["Rutuj", "Rodge", "Yash", "Shloka"]
students.sort()
print(students)
```

```python
### TUPLES 5 OPERATIONS ###
```

```python
students = ("Rutuj", "Yash", "Rodge", "Shloka", "Rutuj", "Aditya", "Rutuj")
print(students.count("Rutuj"))
```

```python
students = ("Rutuj", "Yash", "Rodge", "Shloka", "Aditya")
print(students.index("Rodge"))
```

```python
students = ("Rutuj", "Yash", "Rodge", "Shloka", "Aditya")
print(students[1:4])
```

```python
students = ("Rutuj", "Yash", "Rodge", "Shloka", "Aditya")
others = ("Anirudhha", "Manas", "Lokesh", "Bawankar",)
print(students + others)
```

```python
students = ("Rutuj", "Yash", "Rodge", "Shloka", "Aditya")
print(students * 3)
```

```python
### DICTIONARY 5 OPERATIONS ###
```

```python
students = {
    1: "Rutuj",
    2: "Rodge", 
    3: "Yash",
    4: "Shloka"
}

print(students.keys())
```

```python
students = {
    1: "Rutuj",
    2: "Rodge", 
    3: "Yash",
    4: "Shloka"
}

print(students.values())
```

```python
students = {
    1: "Rutuj",
    2: "Rodge", 
    3: "Yash",
    4: "Shloka"
}

print(students.items())
```

```python
students = {
    1: "Rutuj",
    2: "Rodge", 
    3: "Yash",
    4: "Shloka"
}

print(students.get(3))
print(students.get(0))
print(students.get(1))
print(students.get(2))
print(students.get(4))
```
