```python
import pandas as pd

data = {
    "Roll_No": [1, 2, 3, 3, 4, 5],
    "Name": ["Amit", "Bhavna", "Chetan", "Chetan", "Diya", "Esha"],
    "Marks": [75, None, 68, 68, 90, 72],
    "City": ["Pune", "Mumbai", "Pune", "Pune", "Nashik", None]
}
df = pd.DataFrame(data)

print("Original Data:")
print(df)
print(df.head())
df.info()
print(df.describe())

print("Missing values:")
print(df.isnull().sum())

print("After dropna():")
print(df.dropna())

print("Duplicate rows:")
print(df.duplicated())
df = df.drop_duplicates()

df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
df["City"] = df["City"].fillna(df["City"].mode()[0])

df = df.drop("Roll_No", axis=1)

print("Students with marks above 70:")
print(df.loc[df["Marks"] > 70])
print("Only Name and Marks:")
print(df.loc[:, ["Name", "Marks"]])

print("Cleaned Data:")
print(df)
```
