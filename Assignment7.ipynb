```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Show all rows and columns
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)

# Load the CSV dataset
filename = "Titanic-Dataset.csv"
df = pd.read_csv(filename)

# Display the dataset
print("Dataset:")
print(df)

# Line Plot
plt.figure(figsize=(14, 6))
plt.plot(df["PassengerId"], df["Fare"], color="blue", linewidth=1, label="Fare")
plt.title("Fare of Passengers", fontsize=16)
plt.xlabel("Passenger ID", fontsize=12)
plt.ylabel("Fare", fontsize=12)
plt.grid(True)
plt.legend(fontsize=12)
plt.show()

# Bar Plot
class_counts = df["Pclass"].value_counts().sort_index()
plt.figure(figsize=(10, 6))
bars = plt.bar(class_counts.index.astype(str), class_counts.values, color=["gold", "silver", "peru"], edgecolor="black", label="Passengers")
plt.bar_label(bars, fontsize=12)
plt.title("Passengers in Each Class", fontsize=16)
plt.xlabel("Passenger Class", fontsize=12)
plt.ylabel("Number of Passengers", fontsize=12)
plt.grid(axis="y")
plt.legend(fontsize=12)
plt.show()

# Scatter Plot
plt.figure(figsize=(10, 6))
plt.scatter(df["Age"], df["Fare"], color="green", alpha=0.5, s=40, label="Passengers")
plt.title("Age vs Fare", fontsize=16)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Fare", fontsize=12)
plt.grid(True)
plt.legend(fontsize=12)
plt.show()

# Correlation Matrix and Heatmap
correlation = df[["Survived", "Pclass", "Age", "Fare"]].corr()
print("Correlation Matrix:")
print(correlation)

plt.figure(figsize=(9, 7))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1, linewidths=1, annot_kws={"size": 13})
plt.title("Correlation Heatmap", fontsize=16)
plt.show()

# ----- Extension Task -----

# Histogram of Age
plt.figure(figsize=(10, 6))
plt.hist(df["Age"].dropna(), bins=15, color="skyblue", edgecolor="black", label="Age")
plt.title("Histogram of Age", fontsize=16)
plt.xlabel("Age", fontsize=12)
plt.ylabel("Number of Passengers", fontsize=12)
plt.grid(axis="y")
plt.legend(fontsize=12)
plt.show()

# Box plot of Age and Fare
plt.figure(figsize=(9, 6))
bp = plt.boxplot([df["Age"].dropna(), df["Fare"]], patch_artist=True)
bp["boxes"][0].set_facecolor("lightblue")
bp["boxes"][1].set_facecolor("lightgreen")
plt.xticks([1, 2], ["Age", "Fare"], fontsize=12)
plt.title("Box Plot of Age and Fare", fontsize=16)
plt.xlabel("Column", fontsize=12)
plt.ylabel("Value", fontsize=12)
plt.grid(axis="y")
plt.legend([bp["boxes"][0], bp["boxes"][1]], ["Age", "Fare"], fontsize=12)
plt.show()

# Pie chart of passengers by port
port_counts = df["Embarked"].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(port_counts, labels=port_counts.index, autopct="%1.1f%%", textprops={"fontsize": 13})
plt.title("Passengers by Embarkation Port", fontsize=16)
plt.legend(port_counts.index, title="Port", fontsize=12)
plt.show()

# Average values
print("Average Age:", df["Age"].mean())
print("Average Fare:", df["Fare"].mean())
print("Average Pclass:", df["Pclass"].mean())
```
